```python
import pandas as pd
import numpy as np
```

```python
df1 = pd.read_csv("/content/student_details_batch1.csv")
df2 = pd.read_csv("/content/student_details_batch2.csv")
```

```python
df_master_details = pd.concat([df1, df2],axis=0)
```

```python
df_master_details.head()
```

```python
print(df1.shape)
print(df2.shape)
print(df_master_details.shape)
```

```python
df_master_details['Age'] = df_master_details['Age'].fillna(df_master_details['Age'].mean())
```

```python
df_master_details['Monthly_Allowance'] = df_master_details['Monthly_Allowance'].fillna(df_master_details['Monthly_Allowance'].median())
```

```python
df_master_details['Department'] = df_master_details['Department'].fillna(df_master_details['Department'].mode()[0])
```

```python
df_master_details['Hostel_Status'] = df_master_details['Hostel_Status'].fillna("Not Specified")
```

```python
df_master_details.head()
```

```python
from sklearn.impute import KNNImputer
```

```python
subset_df = df_master_details[['Age', 'Monthly_Allowance','Study_Hours']]
```

```python
imputer = KNNImputer(n_neighbors=5)
```

```python
imputed_data= imputer.fit_transform(subset_df)
```

```python
imputed_df = pd.DataFrame(imputed_data, columns=subset_df.columns)
```

```python
df_master_details['Study_Hours'] = imputed_df['Study_Hours']
```

```python
df = pd.read_csv("/content/student_grades.csv")
```

```python
df_joined = pd.merge(df_master_details, df, on="StudentID", how="inner")
```

```python
df_joined.shape
```

```python
df_left_join = pd.merge(df_master_details, df, on="StudentID", how="left")
df_left_join.shape
```

```python
df_left_join.to_csv("final_student_database.csv",index = False)
```
